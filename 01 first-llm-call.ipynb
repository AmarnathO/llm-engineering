```python
import os
from pathlib import Path

from dotenv import load_dotenv
from openai import OpenAI

# Notebook lives in src/llm_foundation/; .env is at the project root.
env_path = Path.cwd().resolve()
for candidate in [env_path, *env_path.parents]:
    dotenv_file = candidate / ".env"
    if dotenv_file.exists():
        load_dotenv(dotenv_file, override=True)
        break
else:
    raise FileNotFoundError("Could not find .env by walking up from the current directory")

api_key = os.getenv("OLLAMA_API_KEY") or "ollama"
base_url = os.getenv("OLLAMA_HOST") or "http://localhost:11434/v1"
model = os.getenv("OLLAMA_MODEL") or "llama3.2"

print(f"loaded .env from: {dotenv_file}")
print(f"base_url: {base_url}")
print(f"model: {model}")
print(f"api_key present: {bool(api_key)}")

client = OpenAI(api_key=api_key, base_url=base_url)
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[{"role": "user", "content": "Say hello in one sentence."}],
)
print(response.choices[0].message.content)
```
